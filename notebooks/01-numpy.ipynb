## Basics

### Array creation

An array in NumPy is called an ndarray and is multidimensional.
`ndarray` has the following attributes:
- `ndim`: the number of axes/dimensions
- `shape`: the dimensions of the array in a tuple (n,m). The length of the shape tuple is thus `ndim`.
- `size`: the total number of elements in the array. This is equal to the product of the `shape` dimensions.
- `dtype`: the type of the elements in the array. This can be created/specified using standard Python types or NumPy types (int32, float64, complex32, etc.)
- `itemsize`: the size in bytes of each element in the array. So with `dytpe=float64`, `itemsize` will be 64/8=8 bytes. Equivalent to `dytpe.itemsize`.
- `data`: the buffer with the actual elements of the array. However, you typically just access the elements through indexing.

```python
a = np.arange(15).reshape(3,5) # arange(n) creates 1D array consisting of [0-n)
print(f"a = \n{a}")
print(f"shape: {a.shape}")
print(f"ndim: {a.ndim}")
print(f"name: {a.dtype.name}")
print(f"itemsize: {a.itemsize}")
print(f"size: {a.size}")
print(f"type: {type(a)}")
```

```python
# You can also just pass in a regular Python list or tuple into `array()`
a = np.array([2,3,4])
print(f"a = \n{a}")

# `array()` also transforms sequences of sequences (of sequences etc.) into dimensional arrays
b = np.array([(1.5, 2, 3), (4, 5, 6)])
print(f"b = \n{b}")

# Types if inferred, but can be specified
c = np.array([[1, 2], [3, 4]], dtype=np.complex128)
print(f"c = \n{c}")
```

```python
# Often you know the size of the array, but not the contents. To initialize with placeholder values, you can use:
# Initialize with zeros
print(f"zeros\n {np.zeros((3,4))}")
# Initialize with ones (this creates, two 3x4 arrays) with specific type. Default is float64
print(f"\n ones\n {np.ones((2,3,4), dtype=np.int16)}")
# Random values
print(f"\n empty\n {np.empty((2,3))}")
```

```python
# `arange`` creates a sequence of numbers
print(np.arange(15)) # make just a sequential list from 0-n
print(np.arange(10, 30, 5)) # make sequence from a-b jumping by c
print(np.arange(0, 2, 0.3)) # it even accepts FP arguments

# do note that due to FP error, you can run into issues with the number of elements, so it's better to use `linspace`
print(np.linspace(0, 2, 9)) # go from 0-2 with 9 elements
x = np.linspace(0, 2*np.pi, 100) # useful to evaluate lots of points
print(np.sin(x))
```

### Basic Operations

```python
# Arithmetic ops on arrays apply *elementwise*. A new array is created and filled with the result.
a = np.array([20, 30, 40, 50])
b = np.array([0, 1, 2, 3])
c =  a - b
print(c)
d = b**2
print(d)
print(10 * np.sin(a))
print(a < 35)

# The product op * also applies elementwise. This is Hadamard/entrywise product, NOT dot product!
A = np.array([[1, 1],[0, 1]])
B = np.array([[2, 0], [3, 4]])
print(A * B )

# To do dot product, use @ operator or `dot()`
print("\nDot product")
print(A @ B)
print(A.dot(B))
```

```python
# You can apply some operations like += and *= to modify an existing array rather than making a new one
print("\n")
rg = np.random.default_rng(1)  # create instance of default random number generator
a = np.ones((2, 3), dtype=np.int_)
b = rg.random((2, 3))
a *= 3
print(a)
b += a
print(b)
# a += b # ERROR!

# When operating with arrays of different types, the type of the resulting array corresponds
# to the more general or precise one (upcast)
a = np.ones(3, dtype=np.int32)
b = np.linspace(0, np.pi, 3)
print(b.dtype.name)
c = a + b
print(c)
print(c.dtype.name)
d = np.exp(c * 1j)
print(d)
print(d.dtype.name)
```

```python
# Many unary ops like element sum array implemented as methods of the `ndarray` class
a = rg.random((2, 3))
print(a)
print(f"sum = {a.sum()}")
print(f"min = {a.min()}")
print(f"max = {a.max()}")
```

```python
# By default, these ops apply to the array like it's a list of numbers
# You can specify the `axis` parameter to apply an op along a specific axis
b = np.arange(12).reshape(3,4)
print(b)
print(b.sum(axis=0)) # sum of each column
print(b.min(axis=1)) # min of each row
print(b.cumsum(axis=1)) # cumulative sum along each row
```

### Universal functions

```python
# NumPy has familiar functions like sin, cos, exp, which are called "universal functions" (`ufunc`)
# These operate elementwise on an array and produce an array as output.
B = np.arange(3)
print(B)
print(np.exp(B))
print(np.sqrt(B))
C = np.array([2., -1., 4.])
print(np.add(B,C))
```

### Indexing, slicing, and iterating

```python
# 1D arrays can be indexed, sliced, and iterated over like lists and other Pythons sequences
a = np.arange(10)**3
print(a)
print(a[2]) # third element
print(a[2:5]) # third to fifth element
a[:6:2] = 1000
print(a)
print(a[::-1]) # reversed
for i in a:
    print(i**(1/3.))
```

```python
# Multidimensional arrays can have one index per axis. These indices are given in a tuple
def f(x,y):
    return 10 * x + y
b = np.fromfunction(f, (5,4), dtype=np.int_)
print(b)
print(b[2,3]) # (row, col)
print(b[0:5, 1]) # each row in the second column
print(b[:, 1]) # same as previous
print(b[1:3, :]) # each column in the second and third row
# When fewer indices are provided than the number of axes, the missing indices are considered complete slices `:`
print(b[-1]) # the last row, equivalent to b[-1, :]
print("\n")

# The expression `b[i]` is treated as an `i` followed by as many instances of `:` as
# needed to represent the remaining axis.
# You can also write this using dots as `b[i, ...]` which is used to produce as many colons as needed
# for a  complete indexing tuple.
# e.g. for 5 axes, x[1, 2, ...] is equal to x[1, 2, :, :, :] 
c = np.array([[[  0,  1,  2],  # a 3D array (two stacked 2D arrays)
               [ 10, 12, 13]],
              [[100, 101, 102],
               [110, 112, 113]]])
print(c.shape)
print(c[1, ...]) # same as c[1, :, :] or c[1]
print(c[..., 2]) # same as c[:, :, 2]
print("\n")

# Iterating over multidimension arrays is done with respect to the first axis
for row in b:
    print(row)
# But if you want to perform an operation on each element in the array, you can use `.flat` which is an iterator
for element in b.flat:
    print(element)
```

## Shape manipulation

### Changing the shape of an array

```python
# The shape of an array can be changed with various commands.
a = np.floor(10*rg.random((3,4)))
print(f"A = \n{a}")

# The following three commands return a modified array, but do not change the original array.
print(f"Flattened = \n{a.ravel()}") # flattened array
print(f"Reshaped = \n{a.reshape(6,2)}") # modify the shape
print(f"Transposed = \n{a.T}") # transpose array
print(f"{a.shape} vs {a.T.shape}")

# If a dimension is given as -1 in a reshaping operation, the other dimensions are auto calculated
print(f"-1 reshape = \n{a.reshape(3, -1)}")
```

### Stacking together different arrays

```python
# You can stack arrays along different axes
a = np.floor(10 * rg.random((2, 2)))
print(a)
b = np.floor(10 * rg.random((2, 2)))
print(b)
print(f"vstack = \n{np.vstack((a,b))}")
print(f"hstack = \n{np.hstack((a,b))}")

# There's also `column_stack` which stacks 1D arrays as columns into a 2D array. It's like `hstack` only for 2D arrays
from numpy import newaxis
print(f"column_stack = \n{np.column_stack((a,b))}") # with 2d arrays
print()
a = np.array([4., 2.])
b = np.array([3., 8.])
print(f"column_stack = \n{np.column_stack((a,b))}") # Returns a 2D array
print(f"hstack = \n{np.hstack((a,b))}") # see how this is different
print()

print(a[:, newaxis]) # transform a from an array to a 2D column vector
print(np.column_stack((a[:, newaxis], b[:, newaxis])))
print(np.hstack((a[:, newaxis], b[:, newaxis]))) # equivalent result
```

### Splitting one array into several smaller ones

```python
# You can use `hsplit`, `vsplit`, and `arraysplit`
a = np.floor(10 * rg.random((2, 12)))
print(a)
print(np.hsplit(a, 3))
print(np.hsplit(a, (3, 4)))
print(np.vsplit(a, 2))
```

## Copies and views

### No copy at all

```python
# Simple assignments make no copy of objects or their data
a = np.arange(15).reshape(3,5)
b = a   # no new object is created
        # a and b are two names for the same ndarray object
print(b is a) # True

# Python passes mutable objects as references, so function calls make no copy
def f(x):
    print(id(x))
print(id(a)) # same id
f(a)         # same id
```

### View or shallow copy

```python
# Different array objects can share the same data
# `view()` creates a new array object that looks at the same data
c = a.view()
print(c is a) # False
print(c.base is a) # True?, c is a view of the data owned by `a`
                    # docs say True, but this outputs False?
                    # I think it's a little bit of a memory thing, cause it definitely is a view
print(c.flags.owndata) # False
print()

c = c.reshape(5, 3) # `a`'s shape doesn't change, reassigned `c`` is still a view of `a`
print(a.shape) # still (3,5)
c[0, 2] = 1234 # `a`'s data changes
print(a)
print(c)
print()

# Slicing an array returns a view of it
s = a[:, 1:3]
s[:] = 10 # `s[:]` is a view of `s`.
print(a)
```

So essentially you can have two arrays with different shapes but sharing the same data. If you modify one of the arrays, the other array will also be modified.

### Deep copy

```python
# `copy()` forces a deep copy
d = a.copy() # new array object
print(d is a) # False
print(d.base is a) # False
d[0, 0] = 9999
print(a) # Nothing is 9999
print(d)
print()

# Sometimes `copy()` should be called after slicing if the original array is not required anymore.
# e.g. suppose array A is a huge intermediate result and the final result B is only a small fraction of A,
# then a deep copy should be made when constructing B with slicing.
a = np.arange(int(1e8))
b = a[:100].copy()
del a # memory of a can be released
print(b)
# If you didn't do `copy()`, `a` is referenced by `b`b and will thus persist in memory even if you did `del a`
```

## Less basic

### Broadcasting rules

```python
# Broadcasting allows universal functions to deal, in a meaningful way, with inputs that don't have exactly the same shape.
# 1) If all input arrays do not share the same number of dimensions, a "1" will be 
#     repeatedly prepended to the shapes of the similar arrays until all the arrays have the same number of dimensions.
# 2) Ensures that arrays with a size of 1 along a particular dimension act as if they had the size of the array with the 
#     largest shape along that dimension. 
#     The value of the array element is assumed to be the same along that dimension for the "broadcast" array.
# After application of these rules, the size of all arrays must match.
```

## Advanced indexing and index tricks
In addition to indexing by integers and slices, arrays can be indexed by arrays of integers and arrays of booleans.

### Indexing with arrays of indices

```python
a = np.arange(12)**2 # first 12 square numbers
# [1, 4, 9, 16, 25, 36, 49, 64, 81, 100, 121, 144]
i = np.array([1, 1, 3, 8, 5]) # array of indices
print(a[i]) # the elements of `a` at the positions `i`
j = np.array([[3,4], [9,7]]) # bidimensional array of indices
print(a[j]) # same shape as `j`
print()

# When the indexed array `a` is multidimensional, a single array of indices refers to the first dimension.
palette = np.array([[0, 0, 0],         # black
                    [255, 0, 0],       # red
                    [0, 255, 0],       # green
                    [0, 0, 255],       # blue
                    [255, 255, 255]])  # white
image = np.array([[0, 1, 2, 0],  # each value corresponds to a color in the palette
                  [0, 3, 4, 0]])
print(palette[image]) # the (2, 4, 3) color image
print()

# We can also give indexes for more than one dimension. The arrays of indices for each dimension must have the same shape.
a = np.arange(12).reshape(3, 4)
print(a)
i = np.array([[0, 1],  # indices for the first dim of `a`
              [1, 2]])
j = np.array([[2, 1],  # indices for the second dim
              [3, 3]])
print(a[i, j])  # i and j must have equal shape
print(a[i, 2])
print(a[:, j])
```

```python
# In Python, `arr[i, j]` is exactly the same as arr[(i, j)], so we can put `i` and `j` in a tuple and then do indexing with that
l = (i,j)
print(a[l]) # equivalent to `a[i,j]`

# However, we can't do this by putting `i` and `j` into an array, because the array will be interpreted as indexing the dimension of `a``.
s = np.array([i,j]) # not what we want
# print(a[s]) # out of bounds error
print(a[tuple(s)]) # same as `a[i,j]`
```

```python
# Another common use of indexing with arrays is the search of the max value of time-dependent series
time = np.linspace(20, 140, 5) # time scale
data = np.sin(np.arange(20)).reshape(5,4) # time-dependent series
print(time)
print(data)
# index of the maxima for each series
ind = data.argmax(axis=0)
print(ind)
# times corresponding to the maxima
time_max = time[ind]
data_max = data[ind, range(data.shape[1])] # data[ind[0], 0], data[ind[1], 1] ...
print(time_max)
print(data_max)
print(np.all(data_max == data.max(axis=0)))
```

```python
# You can also use indexing with arrays as a target to assign to
a = np.arange(5)
print(a)
a[[1,3,4]] = 0
print(a)
print()
# However, when the list of indices contains repitions, the assignment is done several times, leaving behind the last value
a = np.arange(5)
a[[0, 0, 2]] = [1, 2, 3]
print(a)
print()
# This is reasonable, but watch out if you want to use `+=`, as it may not do what you expect
a = np.arange(5)
a[[0, 0, 2]] += 1
print(a)
# Even though 0 occurs twice in the list of indices, the 0th element is only incremented once.
# This is because Python requires `a += 1` to be equivalent to `a = a + 1`
```
